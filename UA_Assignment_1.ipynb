## Part 2

```python
# Let's describe the url, it is usually easier to do it like this, so in the future, you can easily update the URL
sensors = "https://api.glasgow.gov.uk/traffic/v1/movement/sites"
# Making the query to the web server, using the Get method from the requests library 
response = requests.get(sensors)
response
data = response.json()
```

```python
# empty lists to store what we need
site_ids = []
from_lats = []
from_longs = []
from_descriptions = []

# Iterate over each entry in the JSON array
for entry in data:
    site_id = entry['siteId']
    from_description = entry['from']['description']
    from_lat = entry['from']['lat']
    from_long = entry['from']['long']
    
    # Append values to lists
    site_ids.append(site_id)
    from_descriptions.append(from_description)
    from_lats.append(from_lat)
    from_longs.append(from_long)
    
# Create a Pandas DataFrame, using the lists.
df = pd.DataFrame({
    'siteId': site_ids,
    'from_description': from_descriptions,
    'from_lat': from_lats,
    'from_long': from_longs,
})
```

```python
#Generating sensors geodataframe
gdf_sensors = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df['from_long'], df['from_lat']), crs='EPSG:4326')
gdf_sensors.head()
```

```python
#Loading in the working zones shapefile
Workplace_zones = gpd.read_file('data/WorkplaceZones2011Scotland/WorkplaceZones2011Scotland/WorkplaceZones2011Scotland.shp')
```

```python
#Setting a new crs
gdf_workzones = Workplace_zones.to_crs(epsg=4326)
```

```python
#Exploring the map
gdf_workzones.explore()
```

```python
#Spatial Join of sensors map and working zones
workplace_sensors = gpd.sjoin(gdf_sensors, gdf_workzones, how='inner', predicate='intersects', lsuffix='left', rsuffix='right')
workplace_sensors.head()
```

```python
#Using the glasgow group codes
glasgow = workplace_sensors[workplace_sensors["LADCD"] == "S12000046"]
```

```python
#Grouping the sensors to individual data zones using the groupby function
zoned_sensors = glasgow.groupby("WZCD")["siteId"]

#Creating a separate geodataframe containing only polygons within the Glasgow boundary
glasgow_zones = gdf_workzones[gdf_workzones["LADCD"] == "S12000046"]

#Calculating the no. of sensors per data boundary using the groupby function
zoned_sensors = glasgow.groupby("WZCD").size().reset_index(name="sensors_per_zone")

#Merging the glasgow datazone polygons with the sensors per zone
total = glasgow_zones.merge(zoned_sensors, on="WZCD", how="left")
```

```python
#Plotting the Glasgow working zones with sensors
total.explore()
```

```python
#Creating the final map of data boundaries with more or fewer sensors
total.explore("sensors_per_zone", cmap="PiYG")
```

### Importing datasets for analysis

```python
import requests
import pandas as pd
import folium
import json
import geopandas as gpd
import lonboard
from lonboard import viz
import seaborn as sns
import pydeck as pdk
```

## Part 2 - Retrieving the data using the Api endpoint

```python
#Retrieving New york collisions data using the api endpoint
NY_crashes = "https://data.cityofnewyork.us/resource/h9gi-nx95.json?$limit=50000&$where=latitude IS NOT NULL AND longitude IS NOT NULL"
 
response = requests.get(NY_crashes)
data = response.json()
```

```python
#Converting the data into a pandas dataframe with only the essential columns included
crash_df = pd.DataFrame(data)[
    ["collision_id", "number_of_persons_killed", "number_of_persons_injured", "number_of_cyclist_injured", "number_of_cyclist_killed", "latitude", "longitude"]
].copy()
```

These attributes were included for different reasons: longitude and latitude are required for geospatial analysis, collision_id gives each data point a differentiating attribute unique to them, the rest of the attributes surround casualties of incidents and allow us to gauge the severity of collisions.

```python
#Removing errors in any attribute columns
crash_df = crash_df.apply(pd.to_numeric, errors="coerce")
crash_df.head()
```

```python
#Creating geometry factor to be used to create a Geodataset
geometry = gpd.points_from_xy(crash_df['longitude'], crash_df['latitude'])
```

```python
#Generating the geodataframe for collisions
geo_collisions = gpd.GeoDataFrame(crash_df, geometry=geometry, crs='EPSG:4326')
```

```python
#Using a bounding box of new york to clean up any outlier variables that will mess with data visualisation
import shapely
NYC_bbox = [-75.85,39.91,-72.09,41.67] # New york bounding box
geo_collisions = geo_collisions[geo_collisions.intersects(shapely.box(*NYC_bbox))]
```

## Part 3 + 4 - Mapping the Data, including by variable

```python
#Mapping data with .explore() method
geo_collisions.explore()
```

```python
#Chloropleth mapping data by number_of persons_killed
geo_collisions.explore('number_of_persons_killed', cmap='Blues')
```

```python
#Chloropleth mapping data by number_of_cyclist_killed
geo_collisions.explore('number_of_cyclist_killed', cmap='Oranges')
```

```python
#Visualising data using lonboard
viz(geo_collisions)
```

```python
#Visualising data using the alternative lonboard method
from lonboard import Map, ScatterplotLayer

layer = ScatterplotLayer.from_geopandas(geo_collisions)
map = Map(layers=[layer], height=500)
map
```

```python
#Creating the appropriate pydeck layer for mapping with pydeck
layer = pdk.Layer(
    'HexagonLayer',
    geo_collisions,
    get_position=['longitude', 'latitude'],
    auto_highlight=True,
    elevation_scale=50,
    pickable=True,
    elevation_range=[0, 3000],
    extruded=True,                 
    coverage=1)
```

```python
#Creating the view state for the pydeck map
view_state = pdk.ViewState(
    longitude=-73.970024,
    latitude=40.621790,
    zoom=6,
    min_zoom=5,
    max_zoom=15,
    pitch=40.5,
    bearing=-27.36)
```

```python
#Mapping data using pydeck
r = pdk.Deck(layers=[layer], initial_view_state=view_state)
r.to_html('Collisions.html')
```

### Part 1: You worked with two modern libraries to map big data. Can you describe the differences between working with Longboard and Datashader? Which one provides the most exciting functionality, and how do the outcomes from both of them vary?

Key features of Lonboard: Allows for visualisation of large datasets, vector based.

Key features of Datashader: Allows for visualisation of datasets millions of points in size, raster based.

Key differences: lonboard is more easily imported, requiring no large changes to the environment, datashader however requires environments to be updated.

Furthermore, lonboard visualises entire large datasets as a whole, not changing anything about them when mapping, whereas datashader being a graphics pipeline system, reduces these large datasets to make them more easily usable on regular CPUs. This means lonboard risks slower load times, messier maps with overplotting and the risk of crashing, where datashader plots enough to showcase trends at lightning fast speeds to allow for more elegant spatial data trends to be plotted.

However, lonboard remains more easily interactable, with zooming in and exploring data being quicker than datashaders method.

Most exciting functionality: Datashader due to its lightning fast functionality with enormous datasets and striking visuals like sparks flying in the night.

```python
import zipfile
import os
import pandas as pd
```

```python
#Unzipping the bikes dataset
with zipfile.ZipFile('data/2014-citibike-tripdata.zip', 'r') as zip_ref:
    zip_ref.extractall('data/BikeData')
```

```python
#Compiling all files extracted
csv_files = [os.path.join(root, file) for root, dirs, files in os.walk('data/BikeData') for file in files if file.endswith('.csv')]
```

```python
#Merging all csv's extracted
bike = pd.concat([pd.read_csv(csv_file, encoding="utf-8", encoding_errors="replace") for csv_file in csv_files], ignore_index=True)
#generative AI used for the encoding and encoding_errors command for merging the csv files

bike.head()
```

```python
bike.columns
```

```python
#Removing all other columns
keep_cols = [
    'start station latitude',
    'start station longitude',
]

bike = bike[keep_cols]
bike.head()
```

### Part 4: Use the previous steps and the Datashader documentation to implement a hvplot Map.

```python
#converting latitude and longitude columns to meters for conversion into a datashader map
from datashader.utils import lnglat_to_meters #This instruction was provided by generative AI as there was no instruction to convert longitude and latitude to meters for use by datashader
bike['start station longitude'], bike['start station latitude'] = lnglat_to_meters(
    bike['start station longitude'], bike['start station latitude']
)
```

```python
#Mapping bike data using datashader
import hvplot.pandas

map_tiles = EsriImagery().opts(alpha=0.5, width=900, height=480, bgcolor='black')
plot = bike.hvplot(
    'start station longitude',
    'start station latitude',
    kind='scatter',
    rasterize=True,
    cmap=cc.fire,
    cnorm='eq_hist'
)
map_tiles * plot
```
